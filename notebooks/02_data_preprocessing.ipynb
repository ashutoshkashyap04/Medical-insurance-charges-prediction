```python
import pandas as pd
```

```python
# data loading 

train_df = pd.read_csv("../data/raw/train.csv")
test_df = pd.read_csv("../data/raw/test.csv")

```

```python
# converting the sex column from string to numeric data type - label encoding

# male = 0, female = 1

train_df["is_female"] = (train_df["sex"] == "female").astype(int)
test_df["is_female"] = (test_df["sex"] == "female").astype(int)

train_df.drop(
    columns= ['sex'],
    inplace= True
)

test_df.drop(
    columns= ['sex'],
    inplace= True
)
```

```python
# same for smoker column - smoker = 1, non smoker = 0

train_df["is_smoker"] = (train_df["smoker"] == "yes").astype(int)
test_df["is_smoker"] = (test_df["smoker"] == "yes").astype(int)

train_df.drop(
    columns= ['smoker'],
    inplace= True
)

test_df.drop(
    columns= ['smoker'],
    inplace= True
)
```

```python
# for region column - one hot encoding 

train_df = pd.get_dummies(
    train_df, 
    columns=["region"], 
    dtype=int
)

test_df = pd.get_dummies(
    test_df, 
    columns=["region"], 
    dtype=int
)
```

```python
# Scaling is needed for columns - age, bmi, children

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler.fit(train_df[["age", "bmi", "children"]])

train_df[["age", "bmi", "children"]] = scaler.transform(
    train_df[["age", "bmi", "children"]]
)

test_df[["age", "bmi", "children"]] = scaler.transform(
    test_df[["age", "bmi", "children"]]
)
```

```python
# Pearson correlation calculation

from scipy.stats import pearsonr

selected_features = ['age', 'bmi', 'children', 'is_female', 'is_smoker',
       'region_northeast', 'region_northwest', 'region_southeast',
       'region_southwest'
]

correlations = {
    feature :pearsonr(train_df[feature], train_df['charges'])[0]       # pearsonr returns two values (correlation, p-value)
    for feature in selected_features
}


correlations_df = pd.DataFrame(
    list(correlations.items()),
    columns = ['Feature', 'Pearson Correlation']
)

correlations_df.sort_values(
    by = 'Pearson Correlation',
    ascending = False
)
```

```python
# saving the processed dataset

train_df.to_csv("../data/processed/train.csv", index=False)
test_df.to_csv("../data/processed/test.csv", index=False)
```

```python
# saving the scaler

import joblib

joblib.dump(scaler, "../models/scaler.pkl")
```
